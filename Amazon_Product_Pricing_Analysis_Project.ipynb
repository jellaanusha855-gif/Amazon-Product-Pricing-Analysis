```python
#Outlier
Q3 = df["listed_price"].quantile(0.75)
Q1 = df["listed_price"].quantile(0.25)
IQR = Q3 - Q1
IQR
```

```python
low_limit = Q1 - 1.5*IQR
upp_limit = Q3 + 1.5*IQR
low_limit , upp_limit
```

```python
df = df[(df["listed_price"] >=low_limit) & (df["listed_price"]<=upp_limit)]
```

## EDA

```python
# Scatter Plot
#1. Listed Price vs Selling Price(To compare original price and selling price.)

plt.figure(figsize=(8,6))
plt.scatter(df['listed_price'],df['current/discounted_price'],alpha=0.3)
plt.title('Listed Price vs Selling Price')
plt.xlabel('Listed Price')
plt.ylabel('Selling Price')
plt.show()
```

**Insight:**

The graph shows that products with higher listed prices generally also have higher selling prices. However, the gap between the two prices shows that different products receive different levels of discount.

```python
#2. Pie Chart
#Sponsored vs Organic Products(To show the share of sponsored and organic products.)

sponsored_count = df['is_sponsored'].value_counts()

plt.figure(figsize=(7,7))
plt.pie(sponsored_count,labels=sponsored_count.index,autopct='%1.1f%%',startangle=90)
plt.title('Sponsored vs Organic Products')
plt.show()
```

**Insight:**

The graph shows the share of sponsored and organic products in the dataset. This helps us understand how many products are promoted compared with products that appear normally.

```python
#3.Scatter Plot
#Rating vs Number of Reviews

plt.figure(figsize=(10,5))
plt.scatter(df['rating'],df['number_of_reviews'],alpha=0.5)
plt.title('Rating vs Number of Reviews')
plt.xlabel('Rating')
plt.ylabel('Number of Reviews')
plt.show()
```

**Insight:**

This graph compares product ratings with the number of reviews. Most products have high ratings, mainly between 4 and 5. However, products with the same rating can have very different numbers of reviews. Therefore, a higher rating does not always mean that a product has more reviews.

```python
#4. Barplot
#Average Selling Price by Product Badge(The graph compares the average selling price of products with different product badges)

avg_price = df.groupby('is_best_seller')['current/discounted_price'].mean()
plt.figure(figsize=(10,5))
plt.bar(avg_price.index, avg_price.values)
plt.title('Average Selling Price by Product Badge')
plt.xlabel('Product Badge')
plt.ylabel('Average Selling Price')
plt.xticks(rotation=45)
plt.show()
```

**Insight:**

Products with different badges have different average prices. Products with no badge have a higher average price, while Best Seller products have a lower average price.

```python
#5. Line plot
#Average Price by Rating(The graph shows how the average selling price changes for products with different ratings.)

avg_price_rating = df.groupby('rating')['current/discounted_price'].mean()
plt.figure(figsize=(8,5))
plt.plot(avg_price_rating.index,avg_price_rating.values,marker='o')
plt.title('Average Price by Rating')
plt.xlabel('Rating')
plt.ylabel('Average Selling Price')

plt.show()
```

**Insight:**

This graph shows how the average selling price changes for different product ratings. It helps us understand whether products with higher ratings generally have higher or lower average prices.
